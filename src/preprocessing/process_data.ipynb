```python
import pylidc as pl
import numpy as np
import matplotlib.pyplot as plt
```

```python
scan = pl.query(pl.Scan).first()

print("Paciente:", scan.patient_id)
```

```python
import os
print(os.path.expanduser("~"))
print(os.environ.get("USERPROFILE"))
print(os.environ.get("HOME"))
```

```python
import configparser
configparser.SafeConfigParser = configparser.ConfigParser

vol = scan.to_volume()

print("Shape volumen:", vol.shape)
```

```python
anns = scan.annotations

print("Número de anotaciones:", len(anns))
```

```python
import numpy as np
np.int = int
nodules = scan.cluster_annotations()

print("Número de nódulos:", len(nodules))
```

```python
nod = nodules[0]

print("Radiólogos que lo anotaron:", len(nod))
```

```python
# threshold=0.5 → mayoría de radiólogos
from pylidc.utils import consensus
mask = pl.utils.consensus(nod, clevel=0.5)[0]

print("Shape máscara:", mask.shape)
```

```python
bbox = pl.utils.consensus(nod, clevel=0.5)[1]

print("Bounding box:", bbox)
```

```python
zmin, zmax = bbox[2].start, bbox[2].stop
ymin, ymax = bbox[1].start, bbox[1].stop
xmin, xmax = bbox[0].start, bbox[0].stop

ct_patch = vol[xmin:xmax, ymin:ymax, zmin:zmax]

print("Shape patch CT:", ct_patch.shape)
```
